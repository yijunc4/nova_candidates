# Explore data loading from ALERCE/TAP

```python
from pathlib import Path
ROOT = Path.cwd().parents[0]
```

```python
import pandas as pd
```

```python
import pyvo

# https://science.alerce.online/services/accessing-data-tap/
tap = pyvo.dal.TAPService("https://tap.alerce.online/tap")

query = """
SELECT TOP 100 *
FROM ztf.object
"""

job = tap.search(query)
df_objects = job.to_table().to_pandas()

df_objects.head()
```

```python
df_objects[df_objects.ndet > 1]
```

```python
batch = df_objects["oid"].iloc[50:60].to_list()
objects_sql = ", ".join(f"'{oid}'" for oid in batch)

query = f"""
SELECT *
FROM ztf.detection
WHERE oid IN ({objects_sql})
ORDER BY oid, mjd, candid
"""

job = tap.search(query)
df_detections = job.to_table().to_pandas()

df_detections.head()
```

```python
objects_sql = ", ".join(f"'{oid}'" for oid in batch)

query = f"""
SELECT *
FROM ztf.probability
WHERE oid IN ({objects_sql})
ORDER BY oid
"""

job = tap.search(query)
df_detections = job.to_table().to_pandas()

df_detections.head()
```
